```python
import pymol
import os
import pandas as pd
import numpy as np
from tqdm import tqdm
from rdkit import Chem
from rdkit.Chem import AllChem
```

```python
# 读取目录下的DrugMatrix.csv
df = pd.read_csv('DILI_drugs_oral.csv',encoding='GBK')
df.head()
```

```python
# 设置输出目录为当前文件夹
output_dir = os.getcwd()

# 从SMILES生成RDKit分子对象,SMILES为df['SMILES']
df['mol'] = df['SMILES'].apply(Chem.MolFromSmiles)
df['mol'] = df['mol'].apply(Chem.AddHs)
# 构象优化
# 保存为pdb文件，文件名为CID列的值
# for i in tqdm(range(df.shape[0])):
#     mol = df['mol'][i]
#     cid = df['CID'][i]
#     AllChem.EmbedMolecule(mol,useRandomCoords=True)
#     AllChem.MMFFOptimizeMolecule(mol)
#     Chem.MolToPDBFile(mol, os.path.join(output_dir, f'{cid}.pdb'))

# 修改上面for循环，出现错误或者提示信息时，跳过该分子，继续下一个分子
for i in tqdm(range(df.shape[0])):
    try:
        mol = df['mol'][i]
        cid = df['LTKBID'][i]
        print(f'{cid}.pdb')
        AllChem.EmbedMolecule(mol)
        AllChem.MMFFOptimizeMolecule(mol)
        Chem.MolToPDBFile(mol, os.path.join(output_dir, f'{cid}.pdb'))
    except:
        pass
df.head()
```
